```python


from datetime import datetime

event_title = input("Enter event title: ")
year = int(input("Enter year: "))
month_input = input("Enter month: ").strip()
day = int(input("Enter day: "))

# Convert month name to a number if user entered words
if month_input.isdigit():
    month = int(month_input)
else:
    # Handles both full names ("February") and 3-letter names ("Feb")
    try:
        month = datetime.strptime(month_input, "%B").month
    except ValueError:
        month = datetime.strptime(month_input, "%b").month

# Create and format the date
event_date = datetime(year, month, day)

print("\nFormatted Output:")
print(f"Event: {event_title.upper()}")
print(f"Date: {event_date.strftime('%B')} {day}, {year}")
print(f"Year: {year}")
print(f"Month: {event_date.strftime('%B')}")
print(f"Day: {day}")
print(f"Weekday: {event_date.strftime('%A')}")
```
